```python
import gymnasium as gym
```

```python
env = gym.make("CartPole-v1", render_mode="rgb_array")
env = gym.wrappers.RecordEpisodeStatistics(env)
obs, info = env.reset()
print(obs, info)
G = 0

# terminated means the same thing as "done"
terminated = False
truncated = False
while not terminated or truncated:
  action = env.action_space.sample()
  next_obs, reward, terminated, truncated, info = env.step(action)
  G += reward
  print(next_obs, reward, terminated, truncated, info)

# close environment
env.close()

print(G)
```

```python
def make_env():
  env = gym.make("CartPole-v1", render_mode="rgb_array")
  env = gym.wrappers.RecordEpisodeStatistics(env)
  return env
```

```python
num_envs = 3
envs = gym.vector.SyncVectorEnv(
    [make_env for _ in range(num_envs)])
```

```python
# try with vector env
obs, infos = envs.reset()
print(obs, infos)

# terminated means the same thing as "done"
terminated = False
truncated = False
while not terminated or truncated:
  actions = envs.action_space.sample()
  next_obs, rewards, terminateds, truncateds, infos = envs.step(actions)
  print(next_obs, rewards, terminateds, truncateds, infos)

  # let's stop whenever the first env episode ends
  terminated = terminateds[0]
  truncated = truncateds[0]

# close environment
envs.close()
```
